```python
#word frequency counter 
sent = input("Enter a string sentence :")
words = sent.split()
question = input("Enter a word to count its frequency :")
count = 0
for word in words :
    if question == word :
        count += 1
print("The frequency of the word is :", count)
    

    
```

```python
phone_book = {}
c  = int(input("Enter how many contacts you want to add : "))
for i in range(c):
    name = input("Enter name :")
    num  = int(input("Enter phone number : "))
    phone_book[name] = [num]

for i  in phone_book.items():
    print(i)
```

```python
#comprehensions
evennumbrs = [i for i in range(1,21)if i % 2 == 0 ]
print(evennumbrs)
```

```python
square = {i:i*i for i in range(1,10)}
print(square)
```

```python
sq_odd = {i :i*i for i in range(1,10) if i % 2 != 0}
print(sq_odd)
```
