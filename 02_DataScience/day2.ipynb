```python
import pandas as pd
import matplotlib.pyplot as plt
print("ok")
```

```python
read = pd.read_csv("employee.csv")
print(read)
```

```python
plt.figure(figsize=(4,4))

plt.hist(read["Salary"].dropna())
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("No.of Employees")

plt.show()
```

```python
count = read["Department"].value_counts()
plt.figure(figsize = (3,3))
count.plot(kind = "bar")
plt.title("Bargraph")
plt.xlabel("Department")
plt.ylabel("no.of employess")
plt.show
```

```python
sal = read.groupby("Department")["Salary"].mean()
sal.plot(kind = "bar")

plt.title("Average salary")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()
```

```python
plt.figure(figsize = (2,2))
plt.scatter(read["Experience"],read["Salary"])

plt.title("Scatter plot")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```
